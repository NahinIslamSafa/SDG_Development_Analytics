```python
print("Countries:", all_goals["GeoAreaName"].nunique())
print("Years:", all_goals["TimePeriod"].min(), "to", all_goals["TimePeriod"].max())
print("Indicators:", all_goals["Indicator"].nunique())
print("Series:", all_goals["SeriesCode"].nunique())

print("\nObservations by Goal:")
print(all_goals["Goal"].value_counts().sort_index())

print("\nCountries by Goal:")
print(all_goals.groupby("Goal")["GeoAreaName"].nunique().sort_index())

print("\nYears by Goal:")
print(all_goals.groupby("Goal")["TimePeriod"].agg(["min", "max"]))
```

```python
series_summary = (
    all_goals
    .groupby(["Goal", "SeriesCode", "SeriesDescription"])
    .agg(
        observations=("Value", "size"),
        countries=("GeoAreaName", "nunique"),
        years=("TimePeriod", "nunique"),
        first_year=("TimePeriod", "min"),
        last_year=("TimePeriod", "max"),
        missing_values=("Value_numeric", lambda x: x.isna().sum())
    )
    .reset_index()
)

print(series_summary.to_string(index=False))
```

```python
missing_summary = series_summary[
    series_summary["missing_values"] > 0
].sort_values("missing_values", ascending=False)

missing_summary
```

```python
missing_summary["missing_pct"] = (
    missing_summary["missing_values"]/ missing_summary["observations"] * 100
)

missing_summary
```

```python
missing_by_country = (
    .groupby(["SeriesCode", "GeoAreaName"])
    all_goals[all_goals["Value_numeric"].isna()]
    .size()
    .reset_index(name="missing_values")
    .sort_values(["SeriesCode", "missing_values"], ascending=[True, False])
)

missing_by_country
```

```python
# Dimensions that can represent disaggregation in the SDG dataset
dimensions = [
    "Age",
    "Location",
    "Sex",
    "Education level",
    "Type of skill",
    "Disability status",
    "Hazard type",
    "Quantile"
]

print(dimensions)
```

```python
for dim in dimensions:
    print(f"\n--- {dim} ---")
    print("Missing:", all_goals[dim].isna().sum())
    print("Unique values:", all_goals[dim].nunique(dropna=False))
    print(all_goals[dim].value_counts(dropna=False))
```

```python
dimension_summary = (
    all_goals
    .groupby("SeriesCode")
    .agg(
        age=("Age", lambda x: x.notna().any()),
        location=("Location", lambda x: x.notna().any()),
        sex=("Sex", lambda x: x.notna().any()),
        education=("Education level", lambda x: x.notna().any()),
        skill=("Type of skill", lambda x: x.notna().any()),
        disability=("Disability status", lambda x: x.notna().any()),
        hazard=("Hazard type", lambda x: x.notna().any()),
        quantile=("Quantile", lambda x: x.notna().any())
    )
    .reset_index()
)

dimension_summary
```
